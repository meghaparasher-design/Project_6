```python
# Enable Logging

import logging
import pandas as pd
from pathlib import Path

# --------------------------------------------------
# Create logs folder if it does not exist
# --------------------------------------------------
LOG_DIR = Path("../logs")
LOG_DIR.mkdir(parents=True, exist_ok=True)

LOG_FILE = LOG_DIR / "store_sales_feature_engineering.log"

# --------------------------------------------------
# Create logger
# --------------------------------------------------
logger = logging.getLogger("StoreSales_feature_engineering")
logger.setLevel(logging.DEBUG)

# Prevent duplicate handlers if this cell is run again
if logger.hasHandlers():
    logger.handlers.clear()

# --------------------------------------------------
# Console handler
# --------------------------------------------------
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.DEBUG)

# --------------------------------------------------
# File handler
# --------------------------------------------------
file_handler = logging.FileHandler(
    LOG_FILE,
    mode="a",
    encoding="utf-8"
)
file_handler.setLevel(logging.DEBUG)

# --------------------------------------------------
# Formatter
# --------------------------------------------------
formatter = logging.Formatter(
    "%(asctime)s - %(name)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

# --------------------------------------------------
# Add handlers
# --------------------------------------------------
logger.addHandler(console_handler)
logger.addHandler(file_handler)

# --------------------------------------------------
# First log message
# --------------------------------------------------
logger.info("Store Sales Feature Engineering notebook started.")

print(f"Logging enabled: {LOG_FILE.resolve()}")
```

```python

# EDA AND FEATURE ENGINEERING
logger.info("Starting dataset loading process - store_train and store_test.")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load preprocessed merged datasets
store_train = pd.read_csv("../data/store_train.csv")
store_test = pd.read_csv("../data/store_test.csv")

# Convert Date back to datetime
store_train["Date"] = pd.to_datetime(store_train["Date"])
store_test["Date"] = pd.to_datetime(store_test["Date"])

logger.info("store_train.csv loaded successfully.")
logger.info("store_test.csv loaded successfully.")

# Log dataset dimensions
logger.debug(f"Store Train dataset shape: {store_train.shape}")
logger.debug(f"Store Test dataset shape: {store_test.shape}")

# Check datasets
print("Store Train Shape:", store_train.shape)
print("Store Test Shape:", store_test.shape)

print("\nStore Train:")
print(store_train.head())
print("\nStore Test:")
print(store_test.head())
```

```python
logger.info("store_train.csv loading to convert nan to None.")

store_train = pd.read_csv(
    "../data/store_train.csv",
    keep_default_na=False,
    parse_dates=["Date"],
    dayfirst=True
)
logger.info("store_test.csv loading to convert nan to None.")
store_test = pd.read_csv(
    "../data/store_test.csv",
    keep_default_na=False,
    parse_dates=["Date"],
    dayfirst=True
)
```

```python
print(store_train["PromoInterval"].unique())
print(store_test["PromoInterval"].unique())
```

```python
# CREATE DATE FEATURES
logger.info("Creating Date Features....")
for df in [store_train, store_test]:
    df["Year"] = df["Date"].dt.year
    df["Month"] = df["Date"].dt.month
    df["Day"] = df["Date"].dt.day
    df["WeekOfYear"] = df["Date"].dt.isocalendar().week.astype(int)
    df["DayOfYear"] = df["Date"].dt.dayofyear
    df["Quarter"] = df["Date"].dt.quarter

    # Weekend Feature
    df["IsWeekend"] = df["DayOfWeek"].isin([6, 7]).astype(int)

# COMPETITION OPEN DURATION

for df in [store_train, store_test]:
    df["CompetitionOpenMonths"] = (
        12 * (df["Year"] - df["CompetitionOpenSinceYear"])
        + (df["Month"] - df["CompetitionOpenSinceMonth"])
    )

    # If competition information was missing and filled with 0,
    # avoid unrealistic values
    df.loc[df["CompetitionOpenSinceYear"] == 0,"CompetitionOpenMonths"] = 0

    # Negative values should not exist
    df["CompetitionOpenMonths"] = (
        df["CompetitionOpenMonths"].clip(lower=0)
    )

# PROMO2 ACTIVE DURATION

for df in [store_train, store_test]:
    df["Promo2OpenWeeks"] = (
        (df["Year"] - df["Promo2SinceYear"]) * 52
        + (df["WeekOfYear"] - df["Promo2SinceWeek"])
    )

    df.loc[
        df["Promo2"] == 0,
        "Promo2OpenWeeks"
    ] = 0

    df["Promo2OpenWeeks"] = (
        df["Promo2OpenWeeks"]
        .clip(lower=0)
    )

# PROMO INTERVAL FEATURE

month_map = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun",
    7: "Jul",
    8: "Aug",
    9: "Sept",
    10: "Oct",
    11: "Nov",
    12: "Dec"
}

for df in [store_train, store_test]:
    df["MonthName"] = df["Month"].map(month_map)
    df["IsPromoMonth"] = df.apply(
        lambda row:
        1 if row["MonthName"] in str(row["PromoInterval"]).split(",")
        else 0,
        axis=1
    )

# STATE HOLIDAY CLEANING
logger.info("Cleaning State Holiday....")
for df in [store_train, store_test]:
    df["StateHoliday"] = (
        df["StateHoliday"]
        .astype(str)
        .replace("0.0", "0")
    )

# DROP TEMPORARY FEATURE
logger.info("Droping Temporary Feature....")
store_train.drop(
    columns=["MonthName"],
    inplace=True
)

store_test.drop(
    columns=["MonthName"],
    inplace=True
)


# CHECK FINAL FEATURE ENGINEERED DATA

print("\nFeature Engineered Store Train:")
print(store_train.head())
print("\nFeature Engineered Store Test:")
print(store_test.head())
print("\nStore Train Shape:", store_train.shape)
print("Store Test Shape:", store_test.shape)
print("\nStore Train Columns:")
print(store_train.columns.tolist())

# SAVE FEATURE ENGINEERED DATA
logger.info("Saving store_train to feature_engineered_store_train.csv....")
store_train.to_csv(
    "../data/feature_engineered_store_train.csv",
    index=False
)
logger.info("Saving store_test to feature_engineered_store_test.csv file....")
store_test.to_csv(
    "../data/feature_engineered_store_test.csv",
    index=False
)

print("\nFeature engineered files saved successfully.")
```
